### Plots using ggplot2 and tikz

```r
library(tidyverse)
library(tikzDevice)
```

```r
csv_files = list.files(
  path = "results/simulation_images/", 
  pattern = "*.csv", 
  full.names = TRUE, 
  recursive = TRUE
)
print(csv_files)
```

```r
results = lapply(csv_files, read.csv)
```

#### Experiment 1: Consistency for increasing N

```r
p1 = ggplot(results[[3]] %>% filter(metric == "mspe"), aes(x=n, y=value, group=model)) +
  geom_line(aes(color=model)) +
  geom_point(aes(color=model)) +
  scale_x_log10() +
  labs(
    title="Experiment 1: Consistency for increasing N",
    x="Sample Size N (log scale)",
    y="Mean Squared Prediction Error (MSPE)",
    color="Model"
  ) +
  theme_minimal() +
  theme(
    legend.position="top-right"
  )
p1
```
